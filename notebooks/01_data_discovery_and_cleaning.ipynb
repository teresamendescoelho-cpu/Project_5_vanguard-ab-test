```python
step_order = {
    "start": 0,
    "step_1": 1,
    "step_2": 2,
    "step_3": 3,
    "confirm": 4
}

def has_backward_step(steps):
    positions = [step_order[step] for step in steps]
    return any(positions[i] < positions[i - 1] for i in range(1, len(positions)))

visit_sequences = (
    missing_web_data_sorted
    .groupby("visit_id")["process_step"]
    .apply(list)
)

backward_visits = visit_sequences.apply(has_backward_step)

print("Total visits:", len(visit_sequences))
print("Visits with backward steps:", backward_visits.sum())
```

```python
backward_transitions = []

for steps in visit_sequences:
    positions = [step_order[step] for step in steps]
    
    for i in range(1, len(positions)):
        if positions[i] < positions[i - 1]:
            backward_transitions.append(
                (steps[i - 1], steps[i])
            )

backward_transition_counts = pd.Series(
    backward_transitions
).value_counts()

print("Most common backward transitions:")
print(backward_transition_counts)
```

```python
# Sort data chronologically within each visit
error_data = web_with_variation[
    web_with_variation["Variation"].isin(["Test", "Control"])
].copy()

error_data["date_time"] = pd.to_datetime(error_data["date_time"])

error_data = error_data.sort_values(
    ["visit_id", "date_time"]
)

# Define the order of the process steps
step_order = {
    "start": 0,
    "step_1": 1,
    "step_2": 2,
    "step_3": 3,
    "confirm": 4
}

error_data["step_order"] = error_data["process_step"].map(step_order)

# Previous step within each visit
error_data["previous_step_order"] = (
    error_data.groupby("visit_id")["step_order"].shift(1)
)

# A backward transition occurs when the current step
# is earlier than the previous step
error_data["backward_transition"] = (
    error_data["step_order"] < error_data["previous_step_order"]
)

print(
    error_data["backward_transition"].value_counts(dropna=False)
)
```

# Experiment Evaluation

## 1. Was the experiment well-structured?

The experiment was structured as an A/B test with two groups:

- **Control:** 23,532 clients (46.6%)
- **Test:** 26,968 clients (53.4%)

Both groups were exposed to different versions of the digital process:

- Control → old design
- Test → new design

The experiment ran from **March 15, 2017 to June 20, 2017**, providing several weeks of user interaction data.

The experiment structure was appropriate for comparing the performance of the two designs using completion rate, time spent on each step and error rate.

However, the groups were not perfectly equal in size. The Test group represented 53.4% of the sample while the Control group represented 46.6%. Although this imbalance is not necessarily problematic for an A/B test, it should be considered when evaluating the experiment design.

---

## 2. Were clients randomly and equally divided?

The available data shows that the groups were relatively balanced but not equally divided:

| Group | Clients | Percentage |
|---|---:|---:|
| Control | 23,532 | 46.6% |
| Test | 26,968 | 53.4% |

The data does not provide enough information to independently confirm that clients were randomly assigned to the two groups.

Therefore, randomization cannot be fully verified from the available datasets alone.

The demographic analysis also identified a statistically significant difference in mean age between the groups:

- Control mean age: **47.50 years**
- Test mean age: **47.16 years**
- P-value: **0.0157**

Although statistically significant, the difference is only **0.34 years**, suggesting that the practical impact of this difference is very small.

Gender distribution, on the other hand, did not show a statistically significant difference between the groups:

- Chi-square statistic: **3.6235**
- P-value: **0.3051**

Therefore, gender does not appear to represent a significant imbalance between the Test and Control groups.

---

## 3. Were there potential biases?

Some potential sources of bias should be considered.

First, the Test and Control groups were not exactly equal in size.

Second, the age distribution showed a statistically significant difference between the groups. Although the difference was very small in practical terms, age could potentially influence how clients interact with the digital process.

Another limitation is the presence of clients in the web dataset who were not assigned to either experiment group. These observations were excluded from the main A/B test analysis because they did not have a Test or Control variation.

Therefore, the analysis focuses on clients for whom experiment assignment could be identified.

Other potential biases could include differences in client characteristics, previous digital experience, device type, or other factors that are not available in the provided datasets.

---

## 4. Was the experiment duration adequate?

The experiment ran from **March 15, 2017 to June 20, 2017**, covering approximately three months.

This provides a substantial amount of user interaction data and includes a large number of clients and web interactions.

Based on the available data, the duration appears sufficient to generate meaningful insights into completion rates, time spent on process steps and error rates.

However, without information about expected traffic patterns, seasonality, campaign activity or changes in external conditions during the experiment, it is not possible to determine whether the duration completely controlled for all possible time-related effects.

---

## 5. What additional data would improve the analysis?

The analysis could be strengthened by collecting additional information such as:

- Device type and screen size
- Operating system and browser
- Previous experience with Vanguard's digital platform
- Client digital engagement history
- Traffic source or marketing channel
- Geographic location
- More detailed client segmentation
- Technical errors or system failures
- Page loading times
- Customer support interactions during the experiment
- Revenue or business value generated after completion

This additional information could help identify potential confounding factors and provide a better understanding of why users behave differently between the Test and Control groups.

---

## Overall Experiment Evaluation

Overall, the experiment provides a strong basis for evaluating the new digital design because it contains a large sample of clients and detailed records of their interactions with the digital process.

The main limitations are the unequal group sizes, the inability to independently verify random assignment from the available data, and the small but statistically significant difference in average age between the groups.

These limitations should be considered when interpreting the A/B test results, but they do not invalidate the analysis.
