```python
step_order = {
    "start": 0,
    "step_1": 1,
    "step_2": 2,
    "step_3": 3,
    "confirm": 4
}

def has_backward_step(steps):
    positions = [step_order[step] for step in steps]
    return any(positions[i] < positions[i - 1] for i in range(1, len(positions)))

visit_sequences = (
    missing_web_data_sorted
    .groupby("visit_id")["process_step"]
    .apply(list)
)

backward_visits = visit_sequences.apply(has_backward_step)

print("Total visits:", len(visit_sequences))
print("Visits with backward steps:", backward_visits.sum())
```

```python
backward_transitions = []

for steps in visit_sequences:
    positions = [step_order[step] for step in steps]
    
    for i in range(1, len(positions)):
        if positions[i] < positions[i - 1]:
            backward_transitions.append(
                (steps[i - 1], steps[i])
            )

backward_transition_counts = pd.Series(
    backward_transitions
).value_counts()

print("Most common backward transitions:")
print(backward_transition_counts)
```
